# **Recursion**

**Recursion** is a programming technique in which a function calls itself to solve a problem by breaking it into smaller, simpler subproblems until a specified stopping condition (called the base case) is reached.
Key Components of Recursion
1. Base case: The condition that stops further recursive calls.
2. Recursive case: The part of the function where it calls itself with a smaller or simpler problem.
