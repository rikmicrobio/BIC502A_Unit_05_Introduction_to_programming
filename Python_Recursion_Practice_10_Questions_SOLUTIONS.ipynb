## Q1. Factorial of a Number

Write a recursive function `factorial(n)` to calculate the factorial of a non-negative integer `n`.

**Example:** `factorial(5)` → `120`


```python
def factorial(n):
    # Base case
    if n == 0 or n == 1:
        return 1

    # Recursive case
    return n * factorial(n - 1)

factorial(5)
```

### Explanation
- Base case: when `n` is 0 or 1, return 1.
- Recursive case: multiply `n` by the factorial of `n - 1`.


## Q2. Sum of Natural Numbers

Write a recursive function `sum_n(n)` to calculate the sum of all natural numbers from 1 to `n`.

**Example:** `sum_n(5)` → `15`


```python
def sum_n(n):
    # Base case
    if n == 0:
        return 0

    # Recursive case
    return n + sum_n(n - 1)

sum_n(5)
```

### Explanation
The function adds `n` to the sum of all numbers up to `n - 1`. It stops when `n` reaches 0.


## Q3. Fibonacci Number

Write a recursive function `fibonacci(n)` to calculate the nth Fibonacci number. Assume the sequence starts with 0 and 1.

**Example:** `fibonacci(6)` → `8`


```python
def fibonacci(n):
    # Base cases
    if n == 0:
        return 0
    if n == 1:
        return 1

    # Recursive case
    return fibonacci(n - 1) + fibonacci(n - 2)

fibonacci(6)
```

### Explanation
Each Fibonacci number is the sum of the two previous numbers:
`F(n) = F(n - 1) + F(n - 2)`

The recursion stops at `F(0) = 0` and `F(1) = 1`.


## Q4. Power Calculation

Write a recursive function `power(a, b)` to calculate `a` raised to the power `b`, where `b` is a non-negative integer.

**Example:** `power(2, 4)` → `16`


```python
def power(a, b):
    # Base case
    if b == 0:
        return 1

    # Recursive case
    return a * power(a, b - 1)

power(2, 4)
```

### Explanation
Multiplication by `a` is repeated recursively while the exponent decreases. When the exponent reaches 0, the function returns 1.


## Q5. Count the Number of Digits

Write a recursive function `count_digits(n)` to count the number of digits in a given positive integer.

**Example:** `count_digits(12345)` → `5`


```python
def count_digits(n):
    # Base case
    if n < 10:
        return 1

    # Recursive case
    return 1 + count_digits(n // 10)

count_digits(12345)
```

### Explanation
Integer division by 10 removes the last digit. The function counts one digit each time until only a single digit remains.


## Q6. Sum of Digits

Write a recursive function `sum_digits(n)` to calculate the sum of the digits of a given positive integer.

**Example:** `sum_digits(1234)` → `10`


```python
def sum_digits(n):
    # Base case
    if n == 0:
        return 0

    # Recursive case
    return (n % 10) + sum_digits(n // 10)

sum_digits(1234)
```

### Explanation
- `n % 10` extracts the last digit.
- `n // 10` removes the last digit.
- The function adds each extracted digit until the number becomes 0.


## Q7. Reverse a String

Write a recursive function `reverse_string(s)` to reverse a given string without using loops.

**Example:** `reverse_string('hello')` → `'olleh'`


```python
def reverse_string(s):
    # Base case
    if s == "":
        return ""

    # Recursive case
    return reverse_string(s[1:]) + s[0]

reverse_string("hello")
```

### Explanation
The function recursively reverses the substring from the second character onward, then adds the first character at the end.


## Q8. Palindrome Checker

Write a recursive function `is_palindrome(s)` to determine whether a given string is a palindrome.

**Example:** `is_palindrome('madam')` → `True`


```python
def is_palindrome(s):
    # Base case: empty or single-character strings are palindromes
    if len(s) <= 1:
        return True

    # If the first and last characters differ, it is not a palindrome
    if s[0] != s[-1]:
        return False

    # Recursive case: check the inner substring
    return is_palindrome(s[1:-1])

is_palindrome("madam")
```

### Explanation
The function compares the first and last characters. If they match, it recursively checks the remaining inner substring. If they do not match, it returns `False`.


## Q9. Greatest Common Divisor (GCD)

Write a recursive function `gcd(a, b)` to find the greatest common divisor of two positive integers using the Euclidean algorithm.

**Example:** `gcd(48, 18)` → `6`


```python
def gcd(a, b):
    # Base case
    if b == 0:
        return a

    # Recursive case
    return gcd(b, a % b)

gcd(48, 18)
```

### Explanation
The Euclidean algorithm repeatedly replaces `(a, b)` with `(b, a % b)`. When the remainder becomes 0, the other number is the GCD.


## Q10. Sum of List Elements

Write a recursive function `list_sum(numbers)` to calculate the sum of all elements in a list without using a loop.

**Example:** `list_sum([2, 4, 6, 8])` → `20`


```python
def list_sum(numbers):
    # Base case
    if len(numbers) == 0:
        return 0

    # Recursive case
    return numbers[0] + list_sum(numbers[1:])

list_sum([2, 4, 6, 8])
```

### Explanation
The function adds the first element to the recursive sum of the remaining list. When the list is empty, it returns 0.

## Summary: Recursion

Every recursive function should have:

1. **Base case** — the condition that stops recursion.
2. **Recursive case** — the function calls itself on a smaller or simpler problem.
3. **Progress toward the base case** — each call should move closer to termination.
