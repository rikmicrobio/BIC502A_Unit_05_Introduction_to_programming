## Question 12: Larger number

Write a function named `larger_number` that accepts two numbers and returns the larger number.

### Solution


````python
def larger_number(a, b):
    if a > b:
        return a
    else:
        return b

result = larger_number(12, 8)
print(result)
````

### Expected output

```text
12
```

### Explanation

The function compares `a` and `b` using `if`. If `a` is greater, it returns `a`; otherwise, it returns `b`. The returned value is stored in `result` and displayed using `print()`.

## Question 13: Absolute value

Write a function named `absolute_value` that accepts a number and returns its absolute value without using `abs()`.

### Solution


````python
def absolute_value(number):
    if number < 0:
        return -number
    else:
        return number

print(absolute_value(-7))
print(absolute_value(5))
````

### Expected output

```text
7
5
```

### Explanation

A negative number is converted to its positive form by multiplying it by `-1`. A zero or positive number is returned unchanged. This solution does not use the built-in `abs()` function.

## Question 14: Count characters

Write a function named `count_characters` that accepts a word and returns the number of characters in it.

### Solution


````python
def count_characters(word):
    count = 0
    for character in word:
        count += 1
    return count

print(count_characters("biology"))
````

### Expected output

```text
7
```

### Explanation

The `for` loop visits each character in the word once. The variable `count` increases by 1 for every character. Finally, `return count` sends the total back.

## Question 17: Count vowels

Write a function named `count_vowels` that accepts a word and returns the number of vowels (a, e, i, o, u) in it. Treat uppercase and lowercase letters equally.

### Solution


````python
def count_vowels(word):
    count = 0
    for letter in word.lower():
        if letter in "aeiou":
            count += 1
    return count

print(count_vowels("Molecular Biology"))
````

### Expected output

```text
7
```

### Explanation

The `.lower()` method makes the check case-insensitive. The loop checks each letter, and `if letter in "aeiou"` identifies vowels. The count increases for every vowel found.

## Question 18: Calculate GC content

Write a function named `calculate_gc_content` that accepts a DNA sequence and returns its GC content as a percentage. Assume the sequence contains only A, T, G and C.

### Solution


````python
def calculate_gc_content(sequence):
    gc_count = sequence.count("G") + sequence.count("C")
    total_bases = len(sequence)
    percentage = (gc_count / total_bases) * 100
    return percentage

print(calculate_gc_content("ATGCGC"))
````

### Expected output

```text
66.66666666666666
```

### Explanation

The function counts G and C bases, adds them, and divides this total by the sequence length. Multiplying by 100 converts the fraction into a percentage. In this example, 4 of 6 bases are G or C.

## Question 20: DNA complement

Write a function named `complement_dna` that accepts a DNA sequence and returns its complementary sequence using A–T and G–C pairing.

### Solution


````python
def complement_dna(sequence):
    complement = ""

    for base in sequence:
        if base == "A":
            complement += "T"
        elif base == "T":
            complement += "A"
        elif base == "G":
            complement += "C"
        elif base == "C":
            complement += "G"

    return complement

print(complement_dna("ATGCCG"))
````

### Expected output

```text
TACGGC
```

### Explanation

The loop examines each DNA base. The `if`/`elif` statements add its complementary base to a new string. The completed complementary sequence is returned.

## Question 21: Mean of a list

Write a function named `mean_value` that accepts a list of numbers and returns their arithmetic mean. Assume the list is not empty.

### Solution


````python
def mean_value(values):
    total = 0

    for number in values:
        total += number

    mean = total / len(values)
    return mean

print(mean_value([10, 20, 30, 40]))
````

### Expected output

```text
25.0
```

### Explanation

The loop adds all numbers to `total`. The mean is calculated by dividing the total by the number of items in the list, obtained using `len(values)`. The result is returned.

## Question 22: Largest value in a list

Write a function named `largest_in_list` that accepts a non-empty list of numbers and returns the largest value without using `max()`.

### Solution


````python
def largest_in_list(values):
    largest = values[0]

    for number in values:
        if number > largest:
            largest = number

    return largest

print(largest_in_list([12, 45, 7, 31, 20]))
````

### Expected output

```text
45
```

### Explanation

The function initially assumes that the first item is the largest. It compares every number with the current `largest` value and updates it whenever a bigger number is found. It does not use `max()`.

## Question 24: Filter even numbers

Write a function named `filter_even` that accepts a list of integers and returns a new list containing only the even numbers.

### Solution


````python
def filter_even(values):
    even_numbers = []

    for number in values:
        if number % 2 == 0:
            even_numbers.append(number)

    return even_numbers

print(filter_even([1, 2, 3, 4, 5, 6, 7, 8]))
````

### Expected output

```text
[2, 4, 6, 8]
```

### Explanation

An integer is even when its remainder after division by 2 is zero. The function adds each even number to a new list using `.append()` and returns that list.

## Question 25: Remove spaces

Write a function named `remove_spaces` that accepts a sentence and returns the sentence with all spaces removed.

### Solution


````python
def remove_spaces(sentence):
    result = ""

    for character in sentence:
        if character != " ":
            result += character

    return result

print(remove_spaces("Introduction to Python"))
````

### Expected output

```text
IntroductiontoPython
```

### Explanation

The loop checks each character in the sentence. It adds the character to `result` only if it is not a space. The completed string is returned.

## Question 26: Check a palindrome

Write a function named `is_palindrome` that accepts a word and returns `True` if it reads the same forward and backward, ignoring letter case.

### Solution


````python
def is_palindrome(word):
    word = word.lower()

    if word == word[::-1]:
        return True
    else:
        return False

print(is_palindrome("Level"))
print(is_palindrome("Python"))
````

### Expected output

```text
True
False
```

### Explanation

The word is first converted to lowercase. The slice `word[::-1]` creates the word in reverse order. If the original and reversed words are equal, the function returns `True`; otherwise, it returns `False`.

## Question 27: Factorial using a loop

Write a function named `factorial_value` that accepts a non-negative integer and returns its factorial using a loop. Do not use recursion.

### Solution


````python
def factorial_value(number):
    answer = 1
    i = 1

    while i <= number:
        answer = answer * i
        i += 1

    return answer

print(factorial_value(5))
````

### Expected output

```text
120
```

### Explanation

The variable `answer` starts at 1. The `while` loop multiplies it by each integer from 1 to `number`. For 5, the calculation is 1 × 2 × 3 × 4 × 5 = 120. This solution uses iteration, not recursion.

## Question 28: Temperature status

Write a function named `temperature_status` that accepts a temperature in Celsius and returns `Cold` if it is below 20, `Moderate` if it is from 20 to 30 (inclusive), and `Hot` if it is above 30.

### Solution


````python
def temperature_status(temperature):
    if temperature < 20:
        return "Cold"
    elif temperature <= 30:
        return "Moderate"
    else:
        return "Hot"

print(temperature_status(15))
print(temperature_status(25))
print(temperature_status(35))
````

### Expected output

```text
Cold
Moderate
Hot
```

### Explanation

The first condition checks temperatures below 20. The second condition catches values from 20 through 30 because values below 20 have already been handled. All remaining values are above 30 and return `Hot`.

## Question 29: DNA sequence summary

Write a function named `sequence_summary` that accepts a DNA sequence and returns a dictionary containing the sequence length and the counts of A, T, G and C.

### Solution


````python
def sequence_summary(sequence):
    summary = {
        "length": len(sequence),
        "A": sequence.count("A"),
        "T": sequence.count("T"),
        "G": sequence.count("G"),
        "C": sequence.count("C")
    }

    return summary

print(sequence_summary("ATGCGCAAT"))
````

### Expected output

```text
{'length': 9, 'A': 3, 'T': 2, 'G': 2, 'C': 2}
```

### Explanation

The function creates a dictionary. `len(sequence)` gives the total sequence length, while `.count()` calculates the number of each nucleotide. The dictionary is returned as one value.

## Question 30: Amino-acid frequency

Write a function named `amino_acid_frequency` that accepts a protein sequence and an amino-acid letter, then returns the percentage of residues represented by that letter. Assume the sequence is not empty.

### Solution


````python
def amino_acid_frequency(protein_sequence, amino_acid):
    count = protein_sequence.count(amino_acid)
    total = len(protein_sequence)

    frequency = (count / total) * 100
    return frequency

print(amino_acid_frequency("MKWVTFISLL", "L"))
````

### Expected output

```text
20.0
```

### Explanation

The function counts how many times the requested amino-acid letter occurs, divides that count by the total number of residues, and multiplies by 100. In the example, L occurs 2 times in a 10-residue sequence, so its frequency is 20%.

## Quick Revision: `print()` vs `return`

- `print()` displays information on the screen.
- `return` sends a value from a function back to the statement that called it.
- A returned value can be stored in a variable, used in a calculation, or passed to another function.
- A function can contain conditions and loops before returning its final result.

**Remember:** These solutions use ordinary functions, loops and conditions. No recursive functions are used.
