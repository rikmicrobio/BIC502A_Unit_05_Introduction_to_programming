## Part 1 — Functions

### 1. What is a function?

A **function** is a named, reusable block of code that performs a particular task.

Think of a function as a small machine:
- You give it some input.
- It performs an operation.
- It may give you an output.

**Why use functions?**
- Avoid writing the same code repeatedly.
- Break a large program into smaller tasks.
- Make code easier to read, test, and correct.


````python
# First calculation
a = 10
b = 20
sum = a + b
print(sum)

# Second calculation
a = 15
b = 25
sum = a + b
print(sum)

# Third calculation
a = 30
b = 40
sum = a + b
print(sum)

# Fourth calculation
a = 100
b = 200
sum = a + b
print(sum)
````

### 4. Functions with parameters

A **parameter** is a variable written inside the parentheses in a function definition. It receives an input when the function is called.

An **argument** is the actual value supplied when calling the function.

Example:
- `name` is the parameter.
- `"Riya"` is the argument.


````python
def greet_student(name):
    print("Hello,", name)

greet_student("Riya")
greet_student("Arjun")
````

### 5. Functions with more than one parameter

A function can accept multiple inputs. Separate parameters with commas.


````python
def add_numbers(a, b):
    result = a + b
    print("The sum is:", result)

add_numbers(10, 5)
add_numbers(25, 7)
````

### 6. The `return` statement

The `return` statement sends a result back from a function.

- `print()` displays something on the screen.
- `return` gives a value back to the part of the program that called the function.

Use `return` when you want to store the result, use it in another calculation, or pass it to another function.


````python
def add_numbers(a, b):
    result = a + b
    return result

answer = add_numbers(10, 5)
print("The answer is:", answer)

# We can use the returned value in another calculation:
double_answer = answer * 2
print("Twice the answer is:", double_answer)
````

### 8. Default parameters

A **default parameter** has a value that Python uses when the caller does not provide an argument for it.


````python
def greet(name, message="Welcome to Python!"):
    print(message, name)

greet("Riya")
greet("Arjun", "Good morning,")
````

### 9. Local variables

A **local variable** is created inside a function and is normally available only inside that function.


````python
def calculate_square(number):
    square = number ** 2  # square is a local variable
    return square

print(calculate_square(6))

# Try uncommenting the next line. What happens?
# print(square)
````

### example

Functions are useful for repeating simple calculations in biology. Here, we calculate the percentage of GC bases in a DNA sequence.

GC percentage = (number of G and C bases / total number of bases) × 100


````python
def gc_percentage(sequence):
    sequence = sequence.upper()
    gc_count = sequence.count("G") + sequence.count("C")
    return (gc_count / len(sequence)) * 100

dna = "ATGCGCGTTA"
print("GC percentage:", gc_percentage(dna))
````

**Note:** This introductory example assumes that the DNA sequence is non-empty and contains valid DNA bases. We will add input validation in a later programming lesson.

## Part 2 — Recursion

### 11. What is recursion?

**Recursion** is a programming technique in which a function calls itself to solve a smaller version of the same problem.

Imagine opening nested boxes:
1. Open one box.
2. Find a smaller box inside.
3. Continue until you reach the final box.
4. Then work backwards.

A recursive function needs two essential parts:

1. **Base case:** The condition that stops the function from calling itself.
2. **Recursive case:** The part where the function calls itself with a smaller or simpler input.

Without a suitable base case, recursion may continue until Python raises a `RecursionError`.


### 12. Your first recursive function: countdown

The function prints a number and then calls itself with a number one smaller.

**Trace the calls:** `countdown(3)` → `countdown(2)` → `countdown(1)` → `countdown(0)` → stop.


````python
def countdown(number):
    if number == 0:             # Base case
        print("Done!")
    else:                       # Recursive case
        print(number)
        countdown(number - 1)   # Function calls itself

countdown(3)
````

### 13. Understanding the base case

In the countdown example:
- `if number == 0` is the **base case**.
- `countdown(number - 1)` is the **recursive call**.
- Each call makes the problem smaller by reducing `number`.

### 14. Recursion example: factorial

The factorial of a positive integer is the product of all positive integers up to that number.

For example:

`5! = 5 × 4 × 3 × 2 × 1 = 120`

Mathematically:
- `n! = n × (n-1)!`
- `0! = 1` (base case)


````python
def factorial(n):
    if n == 0:              # Base case
        return 1
    else:                   # Recursive case
        return n * factorial(n - 1)

print("Factorial of 5:", factorial(5))
print("Factorial of 4:", factorial(4))
````

### 15. Trace the factorial function

For `factorial(4)`, Python expands the calls like this:

```text
factorial(4)
= 4 * factorial(3)
= 4 * 3 * factorial(2)
= 4 * 3 * 2 * factorial(1)
= 4 * 3 * 2 * 1 * factorial(0)
= 4 * 3 * 2 * 1 * 1
= 24
```

The function first moves towards the base case. Then the returned values are used as the calls finish.


### 16. Recursion example: sum of numbers

The function below calculates the sum of all integers from `1` to `n`.

For example: `sum_to_n(4) = 4 + 3 + 2 + 1 = 10`.


````python
def sum_to_n(n):
    if n == 0:                 # Base case
        return 0
    return n + sum_to_n(n - 1)

print(sum_to_n(4))
print(sum_to_n(10))
````

### 17. Recursion example: count characters

This example counts the number of characters in a string recursively.

- Base case: an empty string has zero characters.
- Recursive case: count the first character (one) plus the remaining string.


````python
def count_characters(text):
    if text == "":
        return 0
    return 1 + count_characters(text[1:])

print(count_characters("BIOLOGY"))
print(count_characters("PYTHON"))
````

### 18. Function vs recursion

| Feature | Ordinary function | Recursive function |
|---|---|---|
| Meaning | A reusable block of code | A function that calls itself |
| Repetition | Often uses loops or repeated calls | Repeats through self-calls |
| Stopping condition | Depends on the program | Must have a base case |
| Example | `add_numbers(2, 3)` | `factorial(5)` |

**Remember:** Recursion is not always better than a loop. For simple repetition, a loop is often easier and uses less memory. Recursion is useful when a problem can naturally be broken into smaller versions of itself.
